## **3. Operacje na stringach**


```python
# 1. Tworzenie stringów

tekst1 = "Python"
tekst2 = 'Analiza danych'

# 2. Konkatenacja (łączenie stringów)

zdanie = tekst1 + " to świetne narzędzie do " + tekst2
print(zdanie)

# 3. Formatowanie stringów

imie = "Maks"
wiek = 20
saldo = 1523.75
print(f"Mam na imię {imie}, mam {wiek} lat i na koncie {saldo:.2f} PLN.")

# 4. Zaawansowane operacje na stringach

tekst = "Analiza danych w Pythonie jest przydatna."
print(tekst.upper())  # Wszystkie litery na wielkie
print(tekst.lower())  # Wszystkie litery na małe
print(tekst.replace("Pythonie", "R"))  # Zamiana fragmentu tekstu
print(";".join(tekst.split()))  # Łączenie elementów listy
```

# **5. Kolekcje w Pythonie**

```python
# 1. Listy

lista = [10, 20, 30, "tekst", 5.5]
print(lista[0])  # Pierwszy element
lista.append(100)  # Dodanie elementu
lista.remove("tekst")  # Usunięcie elementu

# 2. Krotki (tuples)

krotka = (10, 20, 30)
print(krotka[1])

# 3. Słowniki (dictionaries)

slownik = {"imie": "Anna", "wiek": 28, "saldo": 2500.75}
print(f"{slownik['imie']} ma {slownik['wiek']} lat i {slownik['saldo']:.2f} PLN na koncie.")
```

# **6. Operatory w Pythonie**

```python
# 1. Operatory arytmetyczne

a = 10
b = 3
print(f"Suma: {a + b}, Różnica: {a - b}, Iloczyn: {a * b}, Dzielenie: {a / b:.2f}, Potęga: {a ** b}")

# 2. Operatory porównawcze

print(a > b, a < b, a == b, a != b)

# 3. Operatory logiczne

print((a > 5 and b < 5), (a < 5 or b > 1), not (a < b))

```

# **7. Bloki kodu**

```python
# 1. Instrukcje warunkowe

saldo = 1200
if saldo > 1000:
    print("Masz wystarczające środki.")
elif saldo > 500:
    print("Masz średnie oszczędności.")
else:
    print("Musisz oszczędzać!")

# 2. Pętle

dane = [10, 20, 30, 40]
for liczba in dane:
    print(f"Kwadrat liczby {liczba} wynosi {liczba**2}")

# 3. List Comprehension

kwadraty = [x**2 for x in range(5) if x % 2 == 0]
print(kwadraty)
```

# **8. Funkcje w Pythonie**

```python
def oblicz_vat(cena, vat=0.23):
    return round(cena * (1 + vat), 2)

print(f"Cena z VAT: {oblicz_vat(100)} PLN")
```
