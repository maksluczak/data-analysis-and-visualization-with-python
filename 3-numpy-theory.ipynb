## Broadcasting

Broadcasting to mechanizm NumPy, który pozwala na wykonywanie operacji arytmetycznych na tablicach o różnych wymiarach bez konieczności ich ręcznego dopasowywania.


````python
#Pojedyncza wartość może być rozszerzona (rozprowadzona) na wiele elementów tablicy bez konieczności używania pętli
arr[0:5]=100

#Show
arr
````

````python
# Zresetujmy tablicę.
arr = np.arange(0,25)

#Show
arr
````

````python
#Uwagi dotyczące wycinków
slice_of_arr = arr[0:6]

#Slice
slice_of_arr
````

````python
#Zmiana wycinka
slice_of_arr[:]=99

#Slice
slice_of_arr
````

A teraz wyświetlmy oryginalną tablicę

````python
arr
````

Zauważcie, że zmiany zachodzą również oryginalnej tablicy!

- to nie jest kopia, tylko "widok" na oryginalną tablicę
- dzięki temu oszczędzamy pamięć, ponieważ nie duplikujemy danych

````python
#Aby uzyskać kopię, należy to zrobić jawnie.
arr_copy = arr[0:5].copy()

arr_copy[:]=100
print(arr_copy)
arr
````

### Zaawansowane indeksowanie

Zaawansowane indeksowanie pozwala na wybieranie całych wierszy lub kolumn w dowolnej kolejności. Aby to pokazać, zbudujmy szybko tablicę NumPy:

````python
#Zbuujmy tablicę - na razie 0
arr2d = np.zeros((10,10))
arr2d
````

````python
#Długość wiersza tablicy
arr_length = arr2d.shape[1]
arr_length
````

````python
# Uzupełnienie tablicy

for i in range(arr_length):
    arr2d[i] = i #cały wiersz

arr2d
````

Zaawansowane indeksowanie

````python
arr2d[[2,4,6,8]] # konkretne wiersze - indeks całego wiersza
````

````python
# W odwrotnej kolejności też
arr2d[[6,4,2,7]]
````
