## Pandas - co to w ogóle jest?

## Pandas - kiedy stosować?

## Pandas - struktury danych

- **Series** to **jednowymiarowa** struktura danych (jednowymiarowa macierz
numpy), która oprócz danych przechowuje też unikalny indeks.\
Seria jest bardzo podobna do tablicy NumPy (w rzeczywistości jest zbudowana na obiekcie tablicy NumPy). To, co odróżnia tablicę NumPy od serii, to fakt, że seria może mieć etykiety osi, co oznacza, że ​​może być indeksowana etykietą, a nie tylko lokalizacją liczbową. Nie musi również przechowywać danych liczbowych, może przechowywać dowolny obiekt Pythona.\
Tworzenie: *pd.Series(zawartosc)*

- Drugą strukturą w pandas jest **DataFrame** (ramki danych) - czyli **dwu** lub **więcej wymiarowa** struktura danych, najczęściej w formie tabeli z wierszami i kolumnami.\
  **Kolumny** mają **nazwy**, a **wiersze** mają **indeksy**.\
Tworzenie: *pd.DataFrame(zawartosc)*

## Groupby

Metoda `groupby` pozwala pogrupować wiersze danych razem i wywołać funkcje agregujące.

````python
# Create dataframe
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
       'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
       'Sales':[200,120,340,124,243,350]}
df = pd.DataFrame(data)
df
````

Teraz można użyć metody `.groupby()`, aby grupować wiersze na podstawie nazwy kolumny. Na przykład grupujmy na podstawie Company. Spowoduje to utworzenie obiektu DataFrameGroupBy:

````python
df.groupby('Company')
````

````python
df.groupby("Company")['Sales'].mean()
````

Można zapisać ten obiekt jako nową zmienną

````python
by_comp = df.groupby("Company")['Sales']
````

````python
by_comp
````

### Konkatenacja
Konkatenacja to w zasadzie „sklejanie” ze sobą DataFrame’ów. Pamiętaj, że wymiary muszą się zgadzać wzdłuż osi, po której dokonujesz konkatenacji.
Możesz użyć funkcji pd.concat i przekazać do niej listę DataFrame’ów, które chcesz połąc

````python
pd.concat([df1,df2,df3]) #sklejanie tabel jedna pod drugą
````

````python
pd.concat([df1,df2,df3],axis=1) #sklejanie tabel obok siebie
````

### Łączenie (Merging)
Funkcja `merge` pozwala na łączenie DataFrame’ów przy użyciu logiki podobnej do tej, jaką stosuje się przy łączeniu tabel w SQL.

Nowe przykładowe ramki danych

````python
df1=pd.DataFrame({'animal':['dog', 'cat', 'pig', 'cow'],
                  'age':[5,3,6,2]})
df2=pd.DataFrame({'animal':['dog', 'parrot', 'pig', 'cow', 'horse'],
                  'eat':[2,2,3,5,2]})
````

````python
df1
````

````python
df2
````

````python
pd.merge(df1, df2, on='animal', how='inner')  #część wspólna
````

````python
pd.merge(df1, df2, on='animal', how='left') #do lewej tabeli
````

````python
pd.merge(df1, df2, on='animal', how='right')  #do prawej tabeli
````

````python
pd.merge(df1, df2, on='animal', how='outer')  #wszystkie rekordy z obu tabel
````

````python
pd.merge(df1, df2, on='animal', how='left').dropna()
````

### Łączenie (Joining)
Łączenie to wygodna metoda łączenia kolumn z dwóch DataFrame’ów, które mogą mieć różne indeksy, w jeden wspólny DataFrame jako wynik.

Czyli: scalasz dane po indeksach, a nie po konkretnych kolumnach jak przy merge.

````python
left = pd.DataFrame({'A': ['A0', 'A1', 'A2'],
                     'B': ['B0', 'B1', 'B2']},
                      index=['K0', 'K1', 'K2'])

right = pd.DataFrame({'C': ['C0', 'C2', 'C3'],
                    'D': ['D0', 'D2', 'D3']},
                      index=['K0', 'K2', 'K3'])
````

````python
left
````

````python
right
````

````python
left.join(right)
````

````python
left.join(right, how='outer')
````
